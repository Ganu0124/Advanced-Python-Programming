# 1 Increasing Right Triangle


```python
n = 5
for i in range(1, n + 1):
    for j in range(i):
        print("*", end="")
    print()
```

## 2 Decreasing Right Triangle

```python
n = 5
for i in range(n, 0, -1):
    for j in range(i):
        print("*", end="")
    print()
```

## 3 Right-Aligned Increasing Triangle

```python
n = 5
for i in range(1, n + 1):
    for j in range(n - i):
        print("  ", end="")
    for j in range(i):
        print("* ", end="")
    print()
```

## 4 Right-Aligned Decreasing Triangle

```python
n = 5
for i in range(n, 0, -1):
    for j in range(n - i):
        print("  ", end="")
    for j in range(i):
        print("* ", end="")
    print()
```

## 5 Full Pyramid

```python
n = 5
for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end="")
    
    for j in range(2 * i - 1):
        print("*", end="")
    
    print()
```

## 6 Inverted Pyramid

```python
n = 5
for i in range(n, 0, -1):
    for j in range(n - i):
        print(" ", end="")
    
    for j in range(2 * i - 1):
        print("*", end="")
    
    print()
```

## 7 Diamond

```python
n = 5
for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end="")
    
    for j in range(2 * i - 1):
        print("*", end="")
    print()
for i in range(n - 1, 0, -1):
    for j in range(n - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        print("*", end="")
    print()
```

## 8 Hollow Pyramid

```python
n = 5
for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        if i == n or j == 0 or j == 2 * i - 2:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

# Square Patterns
## 9 Solid Square

```python
n = 5
for i in range(n):
    for j in range(n):
        print("*", end=" ")
    print()
```

## 10 Hollow Square

```python
n = 5
for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

# 12 Same Number Square

# Number Patterns
## 13 Increasing Number Triangle

```python
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

# 14 Decreasing Number Triangle

```python
n = 5
for i in range(n, 0, -1):
    for j in range(1, i + 1):
        print(j, end=" ")
    
    print()
```

# 15 Repeated Number Triangle

```python
n = 5
for i in range(1, n + 1):
    for j in range(i):
        print(i, end=" ")
    print()
```

# 16 Reverse Number Triangle

```python
n = 5
for i in range(1, n + 1):
    for j in range(n, n - i, -1):
        print(j, end=" ")
    
    print()
```

# 17 Continuous Number Triangle

```python
n = 5
num = 1
for i in range(1, n + 1):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()
```

# 18 Number Pyramid

```python
n = 5
for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end=" ")
    for j in range(1, i + 1):
        print(j, end=" ")
    for j in range(i - 1, 0, -1):
        print(j, end=" ")
    
    print()
```

# Alphabet Patterns
## 19 Increasing Alphabet Triangle 

```python
n = 5
for i in range(1, n + 1):
    for j in range(i):
        print(chr(65 + j), end=" ")
    
    print()
```

## 20 Repeated Alphabet Triangle

```python
n = 5
for i in range(n):
    for j in range(i + 1):
        print(chr(65 + i), end=" ")
    
    print()
```

## 21 Continuous Alphabet Triangle

```python
n = 5
letter = 65
for i in range(1, n + 1):
    for j in range(i):
        print(chr(letter), end=" ")
        letter += 1
    
    print()
```

## 22 Reverse Alphabet Triangle

```python
n = 5
for i in range(n, 0, -1):
    for j in range(i):
        print(chr(65 + j), end=" ")
    
    print()
```

# Special Pattern
## 23 X Pattern

```python
n = 5
for i in range(n):
    for j in range(n):
        if j == i or j == n - i - 1:
            print("*", end="")
        else:
            print(" ", end="")
    
    print()
```

## 24 Plus Pattern

```python
n = 5
middle = n // 2
for i in range(n):
    for j in range(n):
        if i == middle or j == middle:
            print("*", end="")
        else:
            print(" ", end="")
    
    print()
```

## 25 Hollow Diamond

```python
n = 5

for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        if j == 0 or j == 2 * i - 2:
            print("*", end="")
        else:
            print(" ", end="")
    
    print()

for i in range(n - 1, 0, -1):
    for j in range(n - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        if j == 0 or j == 2 * i - 2:
            print("*", end="")
        else:
            print(" ", end="")
    
    print()
```

## 26 Hourglass Pattern

```python
n = 5

for i in range(n, 0, -1):
    for j in range(n - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        print("*", end="")
    
    print()

for i in range(2, n + 1):
    for j in range(n - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        print("*", end="")
    
    print()
```

## 27 Filtering Numbers Greater Than 21

```python
numbers = [12, 45, 7, 23, 56, 18, 90]

large_numbers = []

for num in numbers:
    
    if num > 21:
        large_numbers.append(num)

print("Original numbers:", numbers)
print("Numbers greater than 21:", large_numbers)
```

## 28 Reverse a String Using For Loop

```python
original_str = "Python"

reversed_str = ""

for i in range(len(original_str) - 1, -1, -1):
    reversed_str += original_str[i]

print("Original string:", original_str)
print("Reversed string:", reversed_str)
```

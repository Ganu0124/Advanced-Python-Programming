```python
#Factorial series
def factorial(n):
    if n==0:
        return 1
    else:
        return n*factorial(n-1)
print("Factorail of 5:",factorial(5))
```

```python
#Fibonacci series
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)
n = 7
for i in range(n):
    print(fibonacci(i), end=" ")
```

```python
#Sum of n numbers
def sumn(n,num=None):
    if num is None:
        num=[float(input(f"Enter number{i+1}:")) for i in range(n)]
    return sum(num)
n=int(input("Enter n:"))
print(sumn(n))  
```

```python
#Increase every mark by 5 bonus marks
marks=list(map(int,input("Enter marks separated by spaces:").split()))
updated_marks=list(map(lambda x:min(x+5,100),marks))
print("Original Marks:",marks)
print("Updated Marks:",updated_marks)
```

```python
# A University provides a scholarship to student who score at least 80 marks
marks=list(map(int,input("Enter student marks:").split()))
eligible=list(filter(lambda mark:mark>=80,marks))
print("All Marks",marks)
print("Scholarship Eligible:",eligible)
```

```python
names = input("Enter student names: ").split()
marks = list(map(int, input("Enter student marks: ").split()))
students = list(zip(names, marks))
eligible = list(filter(lambda student: student[1] >= 80, students))
print("Scholarship Eligible:")
for name, mark in eligible:
    print(name, mark)
```

```python
# Arranging according to marks in descending order
std=[("A",85),("B",65),("C",95),("D",75)]
std=sorted(std,key=lambda st:st[1],reverse=True)
print("Student sorted by marks")
for st in std:
    print(st)
```

```python
#An online store has a list of product prices
#The system must:
#add 18% gst to every product  
#consider only products costing more than 500
from functools import reduce
pr=list(map(float,input("enter price").split()))
prgst=list(map(lambda x: x*1.18,pr))
filteredpr=list(filter(lambda x: x>500,prgst))
t=reduce(lambda x,y: x+y,filteredpr,0)
print("Original prices",pr)
print("Price with gst",prgst)
print("Poducts above 500:",filteredpr)
print("Total:",round(t,2))
```
